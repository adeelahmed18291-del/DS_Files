```python
# Q1 Stop sending emails after 3 users (Using for loop and break)

users = ["Ali", "Sara", "John", "Zara", "Mike"]
for u in users:
    if u=="Zara":
        print('Limit reached. Stop sending emails Zara')
        break
    else:
        print(f'Email sent to {u}')
```

```python
# Q2. Stop monitoring temperature if critical temperature reached (Using for loop and break)

temps = [22, 25, 30, 45, 28] 
critical = 40
for t in temps:
    if t > critical:
        print(f'Critical temperature reached! Stop monitoring: {t}')
        break
    else:
        print(f'Temperature: {t}')
```

```python
# Q4. Skip customers with unpaid bills (Using For loop and continue)

customers = [ {"name": "Ali", "paid": True},{"name": "Sara", "paid": False},{"name": "John", "paid": True}]
is_not_PaidCustomer = False;
for cust in customers:
    if cust["paid"]==is_not_PaidCustomer:
        continue
        print(cust)
    else:
        print(f'Access granted to:', cust["name"])
```

```python
## Q8. Create a multiplication table using a nested for loop
numbers = [2, 3, 4] 
multipliers = [1, 2, 3]
for o in numbers:
    print(f'Start Here Outer Loop :',o)
    for i in multipliers:
        print(f'Inner Loop : {o} X {i}= {o*i}')
```

```python
## Q9. Keep asking the user for a password until the correct password is entered
correct_pwd = "python123"
password =  input("Correct password is entered: ")
while password != correct_pwd:
    print(f'Enter password: {password} Incorrect password. Try again.')
    password =  input("Correct password is entered: ")
print(f'Enter password:{password} Access granted!') 
```
