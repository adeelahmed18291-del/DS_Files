```python
# Q#.3 Q3. You are managing a list of students enrolled in two different courses. You need to perform various set operations to understand the student enrollment. 
#Task: Create two sets: one for students in "Course A" and one for students in "Course B". Find the students who are in "Course A" but not in "Course B" (difference). 
#Find students who are only in one of the two courses (symmetric difference). 
course_a = {"John", "Alice", "Bob", "David"}
course_b = {"Alice", "Eve", "Charlie", "David"}

difference = course_a.difference(course_b)
print("Difference:", difference)

symmetric_difference = course_a.symmetric_difference(course_b)
print("Symmetric Difference:", symmetric_difference)
```
