```python
# Q3. Write a Python function display_prices that takes a product name
def display_prices(product_name, **prices):
    print(f"Product: {product_name}")
    for store, price in prices.items():
        print(f"{store}: ${price}")
display_prices("Laptop",Amazon=800,eBay=750,Walmart=780)
```

```python
# Q6. Write a Python function calculate_rectangle
def calculate_rectangle(length, width):
    def calculate_area():
        return length * width
    def calculate_perimeter():
        return 2 * (length + width)
    area = calculate_area()  # value move in area variavle
    perimeter = calculate_perimeter()  # value move in perimeter variavle
    return area, perimeter  # Return area, perimeter Values
    result = calculate_rectangle(5, 3)
print("Area:", result[0])
print("Perimeter:", result[1])
```

```python
# Q8. Write a nested function to calculate the total hotel bill
def calculate_hotel_bill(nights):

    def calculate_tax(room_cost):
        return room_cost * 0.10

    room_cost = nights * 5000
    tax = calculate_tax(room_cost)
    total_bill = room_cost + tax

    print("Room Cost:", room_cost)
    print("Tax:", tax)
    print("Total Bill:", total_bill)

calculate_hotel_bill(3)
```

```python
# Q9. Write a Python function Login Validation  (Chatgpt help)
def login():
    # Correct login details
    correct_username = "admin"
    correct_password = "12345"
    # Inner function
    def validate_login(username, password):
        if username == correct_username and password == correct_password:
            return "Login Successful! Welcome."
        else:
            return "Invalid Username or Password!"
    # Take input from user
    username = input("Enter Username: ")
    password = input("Enter Password: ")
    # Call inner function
    result = validate_login(username, password)
    return result
# Call outer function
print(login())
```

```python
# Q10. Write a Python function shopping_bill
def shopping_bill(price, quantity, discount=0):

    # Inner Function 1
    def calculate_total():
        return price * quantity

    # Inner Function 2
    def apply_discount(total):
        return max(0, total - discount)

    # Call inner functions
    total = calculate_total()
    final_bill = apply_discount(total)

    return final_bill


# Example 1: No Discount
print("Final Bill =", shopping_bill(200, 3), "PKR")

# Example 2: With Discount
print("Final Bill =", shopping_bill(200, 3, 100), "PKR")
```
