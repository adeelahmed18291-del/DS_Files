```python
# Q2: Age Group Classification
age = int(input("Enter your age: "))

if age >= 0 and age <= 12:
    print("Child")
elif age >= 13 and age <= 19:
    print("Teenager")
elif age >= 20 and age <= 64:
    print("Adult")
elif age >= 65:
    print("Senior")
else:
    print("Invalid age")
```

```python
# Q6. You are given a list that contains some duplicate items. Remove the duplicates by converting the list to a set. Final output should be in list
shopping_list = ["apple", "banana", "apple", "orange", "banana", "grape"]
print(shopping_list)
shopping_list_set = set(shopping_list) # Assign and Casting
print(shopping_list_set)
```

```python
# Q7. student_details 
student_details = {
    
        "Ali" :{"age":20, "grade":"A","city":"karachi"},
        "Sar" :{"age":21, "grade":"B","city":"Lahore"},
        "Omar" :{"age":19, "grade":"A","city":"Islamabad"},
        "Hina" :{"age":22, "grade":"C","city":"Faisalabad"}
    }

student_details.update({
    "Zara": {
        "age": 20,
        "grade": "B",
        "city": "Multan"
    }
})
print(student_details)
student_details.pop("Omar")
print(student_details)
```

```python
# Q8. inventory
inventory = {
    "The Great Gatsby": {"price": 10.99, "quantity": 5},
    "Rich dad poor dad": {"price": 8.99, "quantity": 8},
    "Harry Potter": {"price": 12.50, "quantity": 2}
}

# 1. Add a new book using update() method
inventory.update({
    "Atomic Habits": {"price": 15.99, "quantity": 4}
})

# 2. Get all book titles using keys() method
print("All Book Titles:")
print(list(inventory.keys()))

inventory["Harry Potter"]["quantity"] = 0  # shirt Stock Change 200 to 250

if inventory["Harry Potter"]["quantity"] == 0:
   inventory.pop("Harry Potter")
   print(list(inventory.keys()))
```

```python
# Q10. program to check whether the entered letter is a vowel, a consonant
letter = input("Enter a single letter: ")

if len(letter) == 1 and letter.isalpha():
    if letter.lower() in "aeiou":
        print(letter, "is a Vowel")
    else:
        print(letter, "is a Consonant")
else:
    print("Please enter valid alphabet")
```
